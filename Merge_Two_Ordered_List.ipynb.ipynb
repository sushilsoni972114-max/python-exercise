```python
def merge_ordered_lists(first, second):
    merged = []
    first_index = 0
    second_index = 0

    while first_index < len(first) and second_index < len(second):
        if first[first_index] <= second[second_index]:
            merged.append(first[first_index])
            first_index += 1
        else:
            merged.append(second[second_index])
            second_index += 1

    merged.extend(first[first_index:])
    merged.extend(second[second_index:])

    return merged


# Test cases
print(merge_ordered_lists([1, 4, 7], [2, 2, 5, 9]))
print(merge_ordered_lists([], [1, 3, 5]))
print(merge_ordered_lists([1, 2, 3], []))
print(merge_ordered_lists([1, 2], [5, 6]))
print(merge_ordered_lists([1, 2, 2, 4], [2, 3, 4]))
```
