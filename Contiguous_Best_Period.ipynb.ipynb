```python
def best_contiguous_period(profits):
    if not profits:
        return None

    best_start = 0
    best_end = 0
    best_total = None

    for start in range(len(profits)):
        for end in range(start, len(profits)):
            period = profits[start:end + 1]
            period_total = sum(period)

            if best_total is None or period_total > best_total:
                best_start = start
                best_end = end
                best_total = period_total

    return {
        "start_index": best_start,
        "end_index": best_end,
        "values": profits[best_start:best_end + 1],
        "total": best_total,
    }


example_profits = [-2, 3, 5, -1, 4, -8, 6]
result = best_contiguous_period(example_profits)

print("Best period:", result)

print("Empty list:", best_contiguous_period([]))
print("All negative:", best_contiguous_period([-8, -3, -10]))
```
