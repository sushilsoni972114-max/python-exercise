```python
temperature_readings = []

while True:
    entry = input("Enter a temperature or 'done' to finish: ").strip()

    if entry.lower() == "done":
        break

    try:
        temperature_readings.append(float(entry))
    except ValueError:
        print("Invalid input. Enter a number or 'done'.")

if not temperature_readings:
    print("No temperature readings were entered.")
else:
    readings_count = len(temperature_readings)
    readings_average = sum(temperature_readings) / readings_count
    readings_above_average = sum(
        reading > readings_average for reading in temperature_readings
    )

    print("Count:", readings_count)
    print("Average:", round(readings_average, 2))
    print("Minimum:", min(temperature_readings))
    print("Maximum:", max(temperature_readings))
    print("Readings above average:", readings_above_average)
```
