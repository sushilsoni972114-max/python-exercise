```python
# Student Scorebook
PASS_MARK = 50
scorebook = []

# Begin with checkpoint 1 and extend the program gradually.


def read_mark(prompt):
    while True:
        try:
            mark = float(input(prompt))
            if 0 <= mark <= 100:
                return mark
            print("Mark must be between 0 and 100.")
        except ValueError:
            print("Enter a valid number.")


def find_student(student_name):
    for row_index, record in enumerate(scorebook):
        if record[0].lower() == student_name.lower():
            return row_index
    return None


def display_record(record):
    student_name = record[0]
    student_marks = record[1:]
    student_total = sum(student_marks)
    student_average = student_total / len(student_marks)
    status = "PASS" if student_average >= PASS_MARK else "FAIL"

    print(
        f"{student_name}: Marks={student_marks}, "
        f"Total={student_total:.2f}, "
        f"Average={student_average:.2f}, {status}"
    )


def add_student():
    student_name = input("Enter student name: ").strip()

    if not student_name:
        print("Name cannot be empty.")
        return

    if find_student(student_name) is not None:
        print("A student with that name already exists.")
        return

    student_marks = [
        read_mark("Enter mark 1: "),
        read_mark("Enter mark 2: "),
        read_mark("Enter mark 3: ")
    ]

    scorebook.append([student_name] + student_marks)
    print("Student added.")


def display_students():
    if not scorebook:
        print("Scorebook is empty.")
        return

    for record in scorebook:
        display_record(record)


def search_student():
    student_name = input("Enter name to search: ").strip()
    student_index = find_student(student_name)

    if student_index is None:
        print("Student not found.")
    else:
        display_record(scorebook[student_index])


def update_mark():
    student_name = input("Enter student name: ").strip()
    student_index = find_student(student_name)

    if student_index is None:
        print("Student not found.")
        return

    try:
        mark_number = int(input("Enter mark number to update (1-3): "))
        if mark_number not in range(1, 4):
            print("Mark number must be 1, 2, or 3.")
            return
    except ValueError:
        print("Enter a valid mark number.")
        return

    scorebook[student_index][mark_number] = read_mark("Enter new mark: ")
    print("Mark updated.")


def remove_student():
    student_name = input("Enter student name to remove: ").strip()
    student_index = find_student(student_name)

    if student_index is None:
        print("Student not found.")
    else:
        removed_record = scorebook.pop(student_index)
        print("Removed:", removed_record[0])


def display_statistics():
    if not scorebook:
        print("Scorebook is empty.")
        return

    all_marks = []
    for record in scorebook:
        all_marks.extend(record[1:])

    print("Number of students:", len(scorebook))
    print("Highest mark:", max(all_marks))
    print("Lowest mark:", min(all_marks))
    print("Overall average:", round(sum(all_marks) / len(all_marks), 2))


def display_ranking():
    if not scorebook:
        print("Scorebook is empty.")
        return

    ranked_records = sorted(
        scorebook,
        key=lambda record: sum(record[1:]) / len(record[1:]),
        reverse=True
    )

    for rank, record in enumerate(ranked_records, start=1):
        print(f"{rank}.", end=" ")
        display_record(record)


while True:
    print("""
1. Add a student
2. Display all students
3. Search for a student
4. Update a mark
5. Remove a student
6. Display class statistics
7. Display ranking
8. Exit
""")

    choice = input("Choose an option: ").strip()

    if choice == "1":
        add_student()
    elif choice == "2":
        display_students()
    elif choice == "3":
        search_student()
    elif choice == "4":
        update_mark()
    elif choice == "5":
        remove_student()
    elif choice == "6":
        display_statistics()
    elif choice == "7":
        display_ranking()
    elif choice == "8":
        print("Exiting scorebook.")
        break
    else:
        print("Invalid choice.")
```
